```python
import os
import matplotlib.pyplot as plt
from PIL import Image

# Defina o caminho da pasta com os arquivos .tif
path_pasta = ""  # Altere para o diretório desejado
```

```python


# Lista todos os arquivos .tif na pasta
arquivos = [f for f in os.listdir(path_pasta) if f.lower().endswith('.tif')]

# Verifica se há imagens na pasta
if not arquivos:
    print("Nenhuma imagem .tif encontrada na pasta.")
else:
    # Define o tamanho da figura e a disposição dos subplots
    num_imgs = len(arquivos)
    cols = 3  # Número de colunas na grade de imagens
    rows = (num_imgs // cols) + (num_imgs % cols > 0)  # Número de linhas
    
    fig, axes = plt.subplots(rows, cols, figsize=(15, 5 * rows))  # Ajuste o tamanho da figura
    axes = axes.flatten()  # Para iterar sobre os eixos mais facilmente
    
    for i, arquivo in enumerate(arquivos):
        img_path = os.path.join(path_pasta, arquivo)
        img = Image.open(img_path)
        
        axes[i].imshow(img, cmap='gray')  # Exibir a imagem em escala de cinza
        axes[i].set_title(arquivo, fontsize=10)
        axes[i].axis("off")  # Remove os eixos para melhor visualização

    # Remove subplots vazios caso o número de imagens não seja múltiplo do número de colunas
    for j in range(i + 1, len(axes)):
        axes[j].axis("off")

    plt.tight_layout()
    plt.show()
```
